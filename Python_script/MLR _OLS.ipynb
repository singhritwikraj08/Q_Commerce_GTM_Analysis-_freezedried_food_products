```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
df = pd.read_excel("Q_Commerce_Data_rating_Count.xlsx")
```

```python
df 
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns
```

```python
target_columns = 'Rating_Count'
feature_columns = ['Quantity', 'MRP', 'Discount',]
```

```python
analysis_df = df[feature_columns+[target_columns]].dropna()
```

```python
for col in analysis_df.columns:
    analysis_df[col] =  pd.to_numeric(analysis_df[col], errors='coerce')
```

```python
analysis_df
```

```python
analysis_df  = analysis_df.dropna()
analysis_df
```

```python
analysis_df.info()
```

```python
x = analysis_df[feature_columns]
y = analysis_df[target_columns]
intercept = sm.add_constant(x)
```

```python
intercept
```

```python
model = sm.OLS(y,intercept).fit()
```

```python
model_summary = model.summary()
```
